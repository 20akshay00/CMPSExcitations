```julia
using CircularCMPS
```

```julia
D = 4;
```

```julia
μ, c = 5., 10.
μ1, μ2, c1, c2, c12 = μ, μ, c, c, c
H1 = MultiBosonLiebLiniger([c1 c12; c12 c2], [μ1, μ2], Inf);
@info "Doing calculation for c1 = $c, c2 = $c, c12 = $c, μ1 = $μ, μ2 = $μ "

ϕ1 = MultiBosonCMPSData_diag(rand, D, 2);
res_d1 = ground_state(H1, ϕ1; gradtol=1e-3, maxiter=1000, do_preconditioning=false);

ψ1 = left_canonical(MultiBosonCMPSData_MDMinv(res_d1[1]))
gs1, _ = ground_state(H1, ψ1; gradtol=1e-10, maxiter=2500, preconditioner_type=2);
```

```julia
H2 = ∫(2 * ∂ψ̂' * ∂ψ̂ - 2 * μ * ψ̂' * ψ̂ + 4 * c * (ψ̂')^2 * ψ̂^2, (-Inf, +Inf));
gs2 = find_groundstate([D], H2, optalg=LBFGS(80; verbosity=1, maxiter=7000, gradtol=1e-10), gradtol=1e-10);
```

```julia
Ra = gs1.M * gs1.Ds[1] / gs1.M
Rb = gs1.M * gs1.Ds[2] / gs1.M;
R1 = [Ra, Rb]

R = gs2.Rs[1][]
R2 = [R, R];
```

```julia
# R1 = R2
c = Rb[1] ./ Ra[1]
γ = (abs2(c) - 2 * c + 1) / (abs2(c) - 2 * conj(c) + 1)
β = 1 / sqrt((abs2((γ + c) / (c - 1)) + 1) * γ)
α = (γ + c) / (c - 1) * β
U = [α β; -conj(β) α]
```

```julia
# R2 = 0
c = Rb[1] ./ Ra[1]
α = 1 / sqrt(1 + abs2(c))
β = conj(c) * α
U = [α β; -conj(β) α]
```

```julia
(U*[Ra, Rb])[1]
```

```julia
H3 = ∫(∂ψ̂' * ∂ψ̂ - μ * ψ̂' * ψ̂ + c * (ψ̂')^2 * ψ̂^2, (-Inf, +Inf));
s = InfiniteCMPS(Constant(Qx), (Constant(Rx),));
expval(H3.h, s)
```
